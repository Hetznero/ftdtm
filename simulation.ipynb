# Agent-Based Model Simulations
This script execute the Fashion Trend Model (FTM), the Demographic Transition Model (DTM), and their fused framework (FTDTM) to simulate fashion trend dynamics with population dynamics

### 1. Fashion Trend Model (FTM)
Simulates trend lifecycles based on agent utility and trend quality.

```python
import numpy as np
from src.ftm import World as FTM

np.random.seed(0)
#initlize world
ftm = FTM(initial_pop=1000, n_initial_trends=5, base_spawn_prob=0.01, total_steps=1200)

#start simulation
print("Starting simulation...")
ftm.run()
print("Simulation complete.")

#Plot results
ftm.plot_results()
ftm.plot_selection_trends([0, 10, 20])
```

### 2. Demographic Transition Model (DTM)
Simulates an aging population by tying mortality and fertility rates to a societal progress curve.

```python
from src.dtm import World as DTM

#initlize world
dtm = DTM(initial_pop=1000, years=300)

#start simulation
print("Starting simulation...")
dtm.run()
print("Simulation complete.")

#plot results
dtm.plot_results()
dtm.plot_resources()

#create gif of population over time
#uncomment/comment if wanted or not
dtm.create_gif('population_pyramid.gif')
```

### 3. Integrated FTMDTM Framework
Couples the models by running fashion dynamics quarterly and demographic updates annually.

```python
from src.ftdtm import World as FTDTM
#initlize world
ftdtm = FTDTM(initial_pop=1000, n_initial_trends=5, years=300, base_spawn_prob=0.01)

#start simulation
print("Starting simulation...")
ftdtm.run()
print("Simulation complete.")

#Plot results
ftdtm.plot_results()
ftdtm.plot_selection_trends([0, 10, 20])
```
